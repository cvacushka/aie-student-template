# Smart Campus AirGuard: EDA и первый эксперимент

Ноутбук показывает базовый сценарий анализа: генерация синтетических данных, проверка распределений, сравнение классов и запуск обучения.

```python
from pathlib import Path
import pandas as pd

from airguard.data.generate_dataset import generate_dataset
from airguard.data.validate import validate_dataset
from airguard.models.train import run_training

data_path = Path('../data/processed/airguard_dataset.csv')
data_path.parent.mkdir(parents=True, exist_ok=True)
df = generate_dataset(rows=6000, seed=42)
df.to_csv(data_path, index=False)
df.head()
```

```python
validate_dataset(df), df.shape, df['co2_risk_next_30min'].mean()
```

```python
df[['co2_ppm_current', 'future_co2_ppm', 'occupancy_count', 'ventilation_level', 'indoor_temp_c', 'humidity_pct']].describe().T
```

```python
df.groupby('co2_risk_next_30min')[['co2_ppm_current', 'occupancy_count', 'ventilation_level', 'indoor_temp_c']].mean().round(2)
```

```python
ax = df['co2_ppm_current'].hist(bins=40, figsize=(8, 4))
ax.set_title('Current CO2 distribution')
ax.set_xlabel('ppm')
```

```python
metrics = run_training(data_path=data_path)
metrics
```
